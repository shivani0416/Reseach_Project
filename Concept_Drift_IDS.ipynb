```python
## cell 14
import os
import numpy as np
import pandas as pd

# -----------------------------
# Paths
# -----------------------------
raw_path = "/content/drive/MyDrive/Concept_Drift_IDS/dataset/raw/MachineLearningCSV/MachineLearningCVE"
processed_path = "/content/drive/MyDrive/Concept_Drift_IDS/dataset/processed"

os.makedirs(processed_path, exist_ok=True)

# -----------------------------
# Get CSV files
# -----------------------------
files = sorted([f for f in os.listdir(raw_path) if f.endswith(".csv")])

print(f"Found {len(files)} CSV files.")

# -----------------------------
# Process each file
# -----------------------------
for file in files:

    print("\n" + "=" * 80)
    print("Processing:", file)

    input_file = os.path.join(raw_path, file)
    output_file = os.path.join(
        processed_path,
        file.replace(".csv", "_processed.csv")
    )

    # Load
    df = pd.read_csv(input_file)

    # Clean column names
    df.columns = df.columns.str.strip()

    # Clean label text
    df["Label"] = df["Label"].astype(str).str.strip()

    # Preserve original attack label
    df["Original_Label"] = df["Label"]

    # Create binary research target
    df["Research_Label"] = (
        df["Label"].str.upper() != "BENIGN"
    ).astype(int)

    # Convert infinite values to NaN
    numeric_columns = df.select_dtypes(include=np.number).columns

    df[numeric_columns] = df[numeric_columns].replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Count missing values before imputation
    missing_before = df[numeric_columns].isna().sum().sum()

    # Median imputation for numerical features
    for column in numeric_columns:
        if df[column].isna().any():
            median_value = df[column].median()
            df[column] = df[column].fillna(median_value)

    # Check remaining missing values
    missing_after = df.isna().sum().sum()

    # Save processed file
    df.to_csv(output_file, index=False)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Missing values handled:", missing_before)
    print("Remaining missing values:", missing_after)
    print("BENIGN:", (df["Research_Label"] == 0).sum())
    print("ATTACK:", (df["Research_Label"] == 1).sum())
    print("Saved:", output_file)

print("\n" + "=" * 80)
print("ALL FILES PROCESSED SUCCESSFULLY!")
```

```python
## cell 17
import os
import pandas as pd

processed_path = "/content/drive/MyDrive/Concept_Drift_IDS/dataset/processed"

# Chronological order: Tuesday → Wednesday → Thursday → Friday
temporal_files = [
    "Tuesday-WorkingHours.pcap_ISCX_processed.csv",
    "Wednesday-workingHours.pcap_ISCX_processed.csv",
    "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX_processed.csv",
    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX_processed.csv",
    "Friday-WorkingHours-Morning.pcap_ISCX_processed.csv",
    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX_processed.csv",
    "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX_processed.csv"
]

dfs = []

for file in temporal_files:
    path = os.path.join(processed_path, file)

    df = pd.read_csv(path)

    dfs.append(df)

    print(f"✅ {file}: {df.shape}")

data = pd.concat(dfs, ignore_index=True)

print("\n===================================")
print("Combined dataset shape:", data.shape)
print("===================================")

print("\nResearch Label distribution:")
print(data["Research_Label"].value_counts())

print("\nResearch Label percentage:")
print((data["Research_Label"].value_counts(normalize=True) * 100).round(2))
```

```python
## cell 35- PSI function
def calculate_psi(reference, comparison, bins=10):
    """
    Calculate Population Stability Index (PSI)
    between a reference distribution and a comparison distribution.
    """

    reference = np.asarray(reference, dtype=float)
    comparison = np.asarray(comparison, dtype=float)

    # Remove invalid values
    reference = reference[np.isfinite(reference)]
    comparison = comparison[np.isfinite(comparison)]

    if len(reference) == 0 or len(comparison) == 0:
        return np.nan

    # Create bins using reference quantiles
    breakpoints = np.percentile(
        reference,
        np.linspace(0, 100, bins + 1)
    )

    # Remove duplicate breakpoints
    breakpoints = np.unique(breakpoints)

    # If feature has no meaningful variation
    if len(breakpoints) < 3:
        return 0.0

    # Expand boundaries
    breakpoints[0] = -np.inf
    breakpoints[-1] = np.inf

    # Calculate histogram proportions
    ref_counts, _ = np.histogram(
        reference,
        bins=breakpoints
    )

    comp_counts, _ = np.histogram(
        comparison,
        bins=breakpoints
    )

    # Convert counts to proportions
    ref_pct = ref_counts / len(reference)
    comp_pct = comp_counts / len(comparison)

    # Avoid division by zero
    epsilon = 1e-6

    ref_pct = np.where(
        ref_pct == 0,
        epsilon,
        ref_pct
    )

    comp_pct = np.where(
        comp_pct == 0,
        epsilon,
        comp_pct
    )

    # PSI formula
    psi = np.sum(
        (comp_pct - ref_pct) *
        np.log(comp_pct / ref_pct)
    )

    return psi
```

```python
## cell 36- Run PSI for all five future periods
future_files = {
    "Thursday WebAttacks":
        "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX_processed.csv",

    "Thursday Infiltration":
        "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX_processed.csv",

    "Friday Morning":
        "Friday-WorkingHours-Morning.pcap_ISCX_processed.csv",

    "Friday PortScan":
        "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX_processed.csv",

    "Friday DDoS":
        "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX_processed.csv"
}

psi_results = []

# Reference sample for efficiency
reference_sample = reference_df.sample(
    n=min(200000, len(reference_df)),
    random_state=42
)

for period, filename in future_files.items():

    print(f"\n🔄 Processing: {period}")

    file_path = os.path.join(
        processed_path,
        filename
    )

    future_df = pd.read_csv(file_path)

    # Sample future data if very large
    future_sample = future_df.sample(
        n=min(100000, len(future_df)),
        random_state=42
    )

    for feature in feature_columns:

        psi_value = calculate_psi(
            reference_sample[feature],
            future_sample[feature],
            bins=10
        )

        psi_results.append({
            "Period": period,
            "Feature": feature,
            "PSI": psi_value
        })

    print(f"✅ Completed: {period}")

psi_df = pd.DataFrame(psi_results)

print("\n🎉 PSI analysis completed!")
print("Shape:", psi_df.shape)
```

```python
## cell 37- Add PSI interpretation
def interpret_psi(value):

    if value < 0.10:
        return "Low"

    elif value < 0.25:
        return "Moderate"

    else:
        return "Substantial"


psi_df["Drift_Level"] = psi_df["PSI"].apply(
    interpret_psi
)

display(
    psi_df.sort_values(
        ["Period", "PSI"],
        ascending=[True, False]
    ).head(20)
)
```

```python
## cell 39-Overall PSI summary
psi_summary = (
    psi_df
    .groupby("Period")
    .agg(
        Mean_PSI=("PSI", "mean"),
        Median_PSI=("PSI", "median"),
        Max_PSI=("PSI", "max"),
        Features_Substantial=(
            "Drift_Level",
            lambda x: (x == "Substantial").sum()
        ),
        Features_Moderate=(
            "Drift_Level",
            lambda x: (x == "Moderate").sum()
        )
    )
    .reset_index()
)

display(psi_summary)
```

```python
# Cell 61B - Create corrected static vs adapted comparison

# Standardize period names in static results
period_mapping = {
    "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX":
        "Thursday WebAttacks",

    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX":
        "Thursday Infiltration",

    "Friday-WorkingHours-Morning.pcap_ISCX":
        "Friday Morning",

    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX":
        "Friday PortScan",

    "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX":
        "Friday DDoS"
}

static_compare = static_results.copy()

static_compare["Period"] = static_compare["Period"].map(
    period_mapping
)

# Keep only relevant metrics
static_compare = static_compare[
    ["Period", "Accuracy", "Precision", "Recall", "F1"]
].rename(
    columns={
        "Accuracy": "Static_Accuracy",
        "Precision": "Static_Precision",
        "Recall": "Static_Recall",
        "F1": "Static_F1"
    }
)

sequential_compare = sequential_results[
    ["Period", "Accuracy", "Precision", "Recall", "F1"]
].rename(
    columns={
        "Accuracy": "Adapted_Accuracy",
        "Precision": "Adapted_Precision",
        "Recall": "Adapted_Recall",
        "F1": "Adapted_F1"
    }
)

# Merge in chronological order
comparison_df = pd.merge(
    static_compare,
    sequential_compare,
    on="Period",
    how="inner"
)

# Calculate changes
comparison_df["Recall_Change"] = (
    comparison_df["Adapted_Recall"]
    - comparison_df["Static_Recall"]
)

comparison_df["F1_Change"] = (
    comparison_df["Adapted_F1"]
    - comparison_df["Static_F1"]
)

print(comparison_df.to_string(index=False))
```
